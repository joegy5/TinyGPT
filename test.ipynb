```python
class T(nn.Module):
    def __init__(self):
        super(T, self).__init__()
        self.p = nn.Linear(3, 3)

    def forward(self, x, u = False):
        if u:
            return self.p(x)
        else:
            return self.p(x)
        
t = T()
i = torch.tensor([[1, 2, 3], [1, 2, 3], [1, 2, 3]], dtype=torch.float32)
o = t(i)
o2 = t(i, u=True)

print(o)
print(o2)
```
